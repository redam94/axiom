# `axiom.infer.Posterior`

`infer` is the sampler seam. The backends (NumPyro, Laplace) arrive in Phase 3; what exists
now is the sampler-free `Posterior` every backend will produce, re-exported here from
`axiom.core.posterior` so that `io` can persist it without importing upward
(note 0002.11).

See `nbs/core/04-protocols-and-posterior.ipynb` for the full tour; this notebook shows the
import path users will keep using once backends exist.

```python
import numpy as np

from axiom.infer import Posterior

rng = np.random.default_rng(0)
post = Posterior({"theta": rng.beta(2, 5, size=(4, 250))}, provenance={"seed": 0, "backend": "none"})
print(post)
print(post.summary("theta", definition="hdi", mass=0.9))
```
